```python
def sum_numbers(nums):
    return sum(nums)
```

```python
def higher_order_functions(f, lst):
    summation = f(lst)
    return summation

higher_order_functions(sum_numbers, [1, 2, 3, 4, 5])
```

```python
def square(x):
    return x ** 2
```

```python
def cube(x):
    return x ** 3
```

```python
def absolute(x):
    if x >= 0:
        return x
    else: return -(x)
```

```python
def higher_order_function(type):
    if type == 'square':
        return square
    elif type == 'cube':
        return cube
    elif type == 'absolute':
        return absolute
```

```python
result = higher_order_function('square')
print(result(3))
result = higher_order_function('cube')
print(result(3))
result = higher_order_function('absolute')
print(result(-3))
```

```python
def add_ten():
    ten = 10
    def add(num):
        return num + ten
    return add

closure_result = add_ten()
print(closure_result(10))
```

```python
def outer_func(msg):
    def inner_func():
        print(msg)
    return inner_func

hi_func = outer_func('Hi')
hello_func = outer_func('Hello')

hi_func()
hello_func()
```

```python
def decorator_func(original_func):
    def wrapper_func():
        print("wrapper_exe.")
        return original_func()
    return wrapper_func

def display():
    print("display is on")

decorated_display = decorator_func(display)
decorated_display()
```

```python
def decorator_func(original_func):
    def wrapper_func(*args, **kwargs):
        print("wrapper_exe.")
        return original_func(*args, **kwargs)
    return wrapper_func

@decorator_func
def display():
    print("display is on")

@decorator_func
def display_info(name, age):
    return f"Name is {name} and age is {age}."

display_info('John', 19)
#display()
```

```python
def uppercase(function):
    def wrapper():
        func = function()
        uppercase = func.upper()
        return uppercase
    return wrapper

def split_string(function):
    def wrapper():
        func = function()
        split = func.split()
        return split
    return wrapper

@split_string
@uppercase
def greeting():
    return "Hello World!"

greeting()
```

```python
def decorator_with_params(function):
    def wrapper(para1, para2, para3):
        function(para1, para2, para3)
        print(f"I live in {para3}.")
    return wrapper

@decorator_with_params
def print_full_name(first, last, country):
    print(f"My name is {first} {last}.")

print_full_name("Hazel", "Lestrange", "UK")
```

```python
# syntax
# map(function, iterable)
```

```python
def square(x):
    return x ** 2

numbers = [1, 2, 3, 4, 5]

numbers_squared = map(square, numbers)
print(list(numbers_squared))
```

```python
numbers_squared = map(lambda x: x ** 2, numbers)
print(list(numbers_squared))
```

```python
numbers_str = ['1', '2', '3', '4', '5']
numbers_int = map(int, numbers_str)
print(list(numbers_int))
```

```python
names = ['Asabeneh', 'Lidiya', 'Ermias', 'Abraham']

def change_to_upper(lst):
    return lst.upper()

names_upper = map(change_to_upper, names)
print(list(names_upper))
```

```python
upper_names = map(lambda names: names.upper(), names)
print(list(upper_names))
```

```python
numbers = [1, 2, 3, 4, 5]

def is_even(num):
    if num % 2 == 0:
        return True
    return False

even = filter(is_even, numbers)
print(list(even))
```

```python
evens = list(filter(lambda n: n % 2 == 0, numbers))
print(evens)
```

```python
numbers = [1, 2, 3, 4, 5]

def is_odd(num):
    if num % 2 != 0:
        return True
    return False

odd = filter(is_odd, numbers)
print(list(odd))
```

```python
names = ['Asabeneh', 'Lidiya', 'Ermias', 'Abraham']

def is_name_long(name):
    if len(name) > 7:
        return True
    return False

long_names = filter(is_name_long, names)
print(list(long_names))
```

```python
from functools import reduce

numbers_str = ['1', '2', '3', '4', '5']

def add_two_nums(x, y):
    return int(x) + int(y)

total = reduce(add_two_nums, numbers_str)
print(total)
```

```python
total = reduce(lambda x,y:int(x)+int(y), numbers_str)
print(total)
```

```python
numbers = [1, 2, 3, 4, 5]
total = reduce(lambda x, y: x + y, numbers)
print(total)
```
