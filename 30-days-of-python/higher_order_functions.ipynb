```python
def square(x):
    return x ** 2
```

```python
def cube(x):
    return x ** 3
```

```python
def absolute(x):
    if x >= 0:
        return x
    else: return -(x)
```

```python
def higher_order_function(type):
    if type == 'square':
        return square
    elif type == 'cube':
        return cube
    elif type == 'absolute':
        return absolute
```

```python
result = higher_order_function('square')
print(result(3))
result = higher_order_function('cube')
print(result(3))
result = higher_order_function('absolute')
print(result(-3))
```

```python
def add_ten():
    ten = 10
    def add(num):
        return num + ten
    return add

closure_result = add_ten()
print(closure_result(10))
```

```python
def outer_func(msg):
    def inner_func():
        print(msg)
    return inner_func

hi_func = outer_func('Hi')
hello_func = outer_func('Hello')

hi_func()
hello_func()
```

```python
def decorator_func(original_func):
    def wrapper_func():
        print("wrapper_exe.")
        return original_func()
    return wrapper_func

def display():
    print("display is on")

decorated_display = decorator_func(display)
decorated_display()
```

```python
def decorator_func(original_func):
    def wrapper_func(*args, **kwargs):
        print("wrapper_exe.")
        return original_func(*args, **kwargs)
    return wrapper_func

@decorator_func
def display():
    print("display is on")

@decorator_func
def display_info(name, age):
    return f"Name is {name} and age is {age}."

display_info('John', 19)
#display()
```

```python
def uppercase(function):
    def wrapper():
        func = function()
        uppercase = func.upper()
        return uppercase
    return wrapper

def split_string(function):
    def wrapper():
        func = function()
        split = func.split()
        return split
    return wrapper

@split_string
@uppercase
def greeting():
    return "Hello World!"

greeting()
```

```python
def decorator_with_params(function):
    def wrapper(para1, para2, para3):
        function(para1, para2, para3)
        print(f"I live in {para3}.")
    return wrapper

@decorator_with_params
def print_full_name(first, last, country):
    print(f"My name is {first} {last}.")

print_full_name("Hazel", "Lestrange", "UK")
```

```python
countries = ['Estonia', 'Finland', 'Sweden', 'Denmark', 'Norway', 'Iceland']

def count_by_starting_letter(countries):
    country_dict = {}

    for country in countries:
        first_letter = country[0]

        if first_letter in country_dict:
            country_dict[first_letter] += 1
        else:
            country_dict[first_letter] = 1

    return country_dict

print(count_by_starting_letter(countries))
```
